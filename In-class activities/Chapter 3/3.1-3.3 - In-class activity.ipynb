## Quick Recap

- **Logistic regression** predicts probabilities using the sigmoid function — output is P(class = 1)
- By default, predictions use a **threshold of 0.5**: if P(class = 1) > 0.5, predict 1
- **Accuracy** = (TP + TN) / total — but can be misleading when classes are imbalanced
- **True Positive Rate (TPR)** = TP / (TP + FN) — of all the actual positives, how many did we catch?
- **True Negative Rate (TNR)** = TN / (TN + FP) — of all the actual negatives, how many did we correctly identify?

### 5. Regularization in logistic regression

In Chapter 2, you used `alpha` to control regularization strength for Ridge and Lasso. Logistic regression has the same idea, but with a different convention:

- The parameter is called `C`, and it's the **inverse** of regularization strength: **`C = 1 / alpha`**
- So **smaller C = stronger regularization** (more shrinkage), **larger C = weaker regularization** (less shrinkage)
- The `penalty` parameter controls the type: `'l2'` (Ridge-like, default) or `'l1'` (Lasso-like, requires `solver='liblinear'`)
